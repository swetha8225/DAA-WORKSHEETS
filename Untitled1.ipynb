```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('ggplot')
```

```python
from google.colab import files
uploaded=files.upload()
df=pd.read_csv("sample_titanic.csv")
df.head()
```

```python
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Distribution")
plt.show()
print(df['Survived'].value_counts())
print(df['Survived'].value_counts(normalize=True)*100)
```

```python
sns.countplot(x='Sex', data=df)
plt.show()
print(df['Sex'].value_counts())
```

```python
sns.countplot(x='Pclass',data=df)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'],bins=30,kde=True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Fare'],bins=40,kde=True)
plt.show()
```

```python
print(df.isnull().sum())
sns.heatmap(df.isnull(),cbar=False)
plt.show()
```

```python
sns.countplot(x='Sex',hue='Survived',data=df)
plt.show()
```

```python
sns.countplot(x='Pclass',hue='Survived',data=df)
plt.show()
print(pd.crosstab(df['Pclass'],df['Survived']))
```

```python
sns.countplot(x='Embarked',hue='Survived',data=df)
plt.show()
```

```python
sns.boxplot(x=df['Age'])
plt.show()
```

```python
sns.boxplot(x=df['Fare'])
plt.show()
```

```python
numeric_df=df.select_dtypes(include=np.number)
plt.figure(figsize=(10,6))
sns.heatmap(numeric_df.corr(),annot=True,cmap='coolwarm')
plt.show()
```

```python
sns.pairplot(df[['Age','Fare','Pclass','Survived']])
plt.show()
```

```python
sns.boxplot(x='Survived',y='Age',data=df)
plt.show()
```

```python
sns.boxplot(x='Survived',y='Fare',data=df)
plt.show()
```

```python
df['FamilySize']=df['SibSp']+df['Parch']+1
sns.countplot(x='FamilySize',data=df)
plt.show()
print(df.groupby('FamilySize')['Survived'].mean())
```

```python
bins=[0,16,60,100]
labels=['Child','Adult','Senior']
df['AgeGroup']=pd.cut(df['Age'],bins=bins,labels=labels)
sns.countplot(x='AgeGroup',hue='Survived',data=df)
plt.show()
```

```python
df['FareGroup']=pd.qcut(df['Fare'],q=3,labels=['Low','Medium','High'])
sns.countplot(x='FareGroup',hue='Survived',data=df)
plt.show()
```

```python
print(df.groupby('Pclass')['Age'].mean())
sns.barplot(x='Pclass',y='Age',data=df)
plt.show()
```
